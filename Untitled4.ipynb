```python
billAmount = int(input("What is the total bill?"))

if billAmount >= 25000:
    discount = billAmount * 0.15
elif billAmount >= 15000:
    discount = billAmount * 0.10
elif billAmount >= 10000:
    discount = billAmount * 0.05
elif billAmount >= 5000:
    discount = billAmount * 0.02
else:
    discount = billAmount * 0.0

print("Total", billAmount)
print("Discount", discount)
print("Payable", billAmount - discount)
```
